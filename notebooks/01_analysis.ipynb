# 客服会话全量质检

## 1. 加载数据

```python
import pandas as pd
pd.set_option('display.width', 120)
df = pd.read_csv('data/dataset.csv')
print(df.shape)
df.head()
```

## 2. 自动质检规则与准确性

```python
pred = ((df.keyword_flag==1)|(df.emotion<0.4)).astype(int)
print(pd.crosstab(df.label_bad, pred))
```

## 3. 分阶段差评率

```python
df.groupby('period').label_bad.mean()
```
